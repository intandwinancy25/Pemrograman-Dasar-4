```python
# ==================================================
# Praktikum 1 - Input dan Output
# Menerima Input Tipe Data Float
# ==================================================

# Nama    : Intan Dwinancy Aprilia Nainggolan
# NIM     : 4232501035
# Tujuan  : Menerapkan fungsi input() dan konversi tipe data
#           float() pada program Python.

total = float(input("Berapa porsi makan yang Anda beli? : "))
nama = str(input("Siapa Nama Pelanggan? : "))

print("Nama Pelanggan :", nama)
print("Jumlah Porsi   :", total)
```

```python
# ==================================================
# Praktikum 2 - Input dan Output
# Praktik 2: Kalkulator Sederhana
# ==================================================

# Nama    : Intan Dwinancy Aprilia Naiggolan
# NIM     : 4232501035
# Tujuan  : Menerapkan input, variabel, dan operator
#           aritmetika untuk melakukan perhitungan.

a = float(input("Masukkan bilangan pertama : "))
b = float(input("Masukkan bilangan kedua   : "))

print("Penjumlahan :", a + b)
print("Pengurangan :", a - b)
print("Perkalian   :", a * b)
print("Pembagian    :", a / b)
```

```python
# ==================================================
# Praktikum 3 - Input dan Output
# Praktik 3: Kalkulator Kasir
# ==================================================

# Nama    : Intan Dwinancy Aprilia Nainggolan
# NIM     : 4232501035
# Tujuan  : Menggunakan input dan operator perkalian
#           untuk menghitung total harga pembelian.

harga_makan = int(input("Berapa harga mi ayam per porsi? : "))
jumlah_makan = int(input("Berapa porsi yang Anda beli?     : "))

total_makan = harga_makan * jumlah_makan

print(f"Harga makan per porsi = {harga_makan}, "
      f"totalnya adalah {total_makan}")
```

```python
# ==================================================
# Praktikum 2 - Input dan Output
# Praktik 4: Mengecek Bilangan
# ==================================================

# Nama    : Intan Dwinancy Aprilia Nainggolan
# NIM     : 4232501035
# Tujuan  : Menerapkan operator aritmetika dan logika
#           untuk mengecek kondisi suatu bilangan.

bilangan = int(input("Masukkan sebuah bilangan : "))

hasil = (bilangan > 0) and (bilangan % 2 == 0)

print("Bilangan genap dan positif? :", hasil)
```

```python
# ==================================================
# Praktikum 1 - Ekspresi
# Konversi Total Detik
# ==================================================

# Nama    : Intan Dwinancy Aprilia Nainggolan
# NIM     : 4232501035
# Tujuan  : Menerapkan operator pembagian bulat (//)
#           dan modulus (%) untuk mengubah total detik
#           menjadi jam, menit, dan detik.

total_detik = int(input("Masukkan total detik : "))

jam = total_detik // 3600
sisa_detik = total_detik % 3600
menit = sisa_detik // 60
detik = sisa_detik % 60

print(f"Hasil Konversi: {jam} Jam, {menit} Menit, {detik} Detik")
```

```python
# ==================================================
# Praktikum 2 - Ekspresi
# Menghitung Nilai Akhir Mahasiswa
# ==================================================

# Nama    : Intan Dwinancy Aprilia Nainggolan
# NIM     : 4232501035
# Tujuan  : Menerapkan ekspresi aritmetika dan relasional
#           untuk menghitung nilai akhir serta menentukan
#           status kelulusan mahasiswa.

tugas = float(input("Masukkan nilai tugas : "))
uts = float(input("Masukkan nilai UTS   : "))
uas = float(input("Masukkan nilai UAS   : "))

standar_kelulusan = float(input("Masukkan standar kelulusan : "))

nilai_mahasiswa = (0.3 * tugas) + (0.3 * uts) + (0.4 * uas)

apakah_lulus = nilai_mahasiswa >= standar_kelulusan
perlu_remedial = nilai_mahasiswa < standar_kelulusan

print("\n===== HASIL =====")
print(f"Nilai Mahasiswa   : {nilai_mahasiswa}")
print(f"Standar Kelulusan : {standar_kelulusan}")
print(f"Apakah Mahasiswa Lulus? : {apakah_lulus}")
print(f"Perlu Remedial?          : {perlu_remedial}")
```

```python
# ==================================================
# Praktikum 3 - Ekspresi
# Menghitung Diskon Belanja
# ==================================================

# Nama    : Intan Dwinancy Aprilia Nainggolan
# NIM     : 4232501035
# Tujuan  : Menerapkan ekspresi aritmetika dan relasional
#           untuk menghitung diskon berdasarkan total
#           belanja.

belanja = float(input("Masukkan total belanja : "))

diskon = (0, 0.35 * belanja)[belanja > 1000000]

total_harga = belanja - diskon

print("\n===== HASIL =====")
print(f"Total Belanja : Rp{belanja:,.0f}")
print(f"Diskon        : Rp{diskon:,.0f}")
print(f"Total Harga   : Rp{total_harga:,.0f}")
```

```python
# ==================================================
# Latihan 1 - Menentukan Tahun Kabisat
# ==================================================

# Nama    : Intan Dwinancy Aprilia Nainggolan
# NIM     : 4232501035
# Tujuan  : Menerapkan operator aritmetika, relasional,
#           dan logika untuk menentukan tahun kabisat.

tahun = int(input("Masukkan tahun : "))

tahun_kabisat = (tahun % 400 == 0) or (
    tahun % 4 == 0 and tahun % 100 != 0
)

print(f"Tahun {tahun} merupakan tahun kabisat? : {tahun_kabisat}")
```

```python
# ==================================================
# Latihan 2 - Program Sederhana
# Menghitung Total Harga Barang
# ==================================================

# Nama    : Intan Dwinancy Aprilia Nainggolan
# NIM     : 4232501035
# Tujuan  : Menerapkan fungsi input(), output(),
#           variabel, dan operator aritmetika dalam
#           sebuah program sederhana.

nama_barang = input("Masukkan nama barang : ")
harga = float(input("Masukkan harga barang : "))
jumlah = int(input("Masukkan jumlah barang : "))

total = harga * jumlah

print("\n===== HASIL PEMBELIAN =====")
print("Nama Barang :", nama_barang)
print("Harga       :", harga)
print("Jumlah      :", jumlah)
print("Total Harga :", total)
```
